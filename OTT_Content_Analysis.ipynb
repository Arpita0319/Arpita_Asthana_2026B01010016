```python
import pandas as pd
import numpy as np
print("Libraries imported successfully!")
```

```python
df = pd.read_csv("/content/netflix_titles[1].csv")

print("Dataset uploaded successfully!")

print("First five records:")
display(df.head())
```

```python
# Displaying the last five records

print("Last five records:")
display(df.tail())
```

```python
# Checking the number of rows and columns

print("Dataset shape:", df.shape)

# Displaying all column names

print("Dataset columns:")
print(df.columns.tolist())

# Displaying dataset information

print("\nDataset information:")
df.info()
```

```python
# Displaying statistical information

print("Statistical description:")
display(df.describe())

# Displaying description of all columns

print("Complete dataset description:")
display(df.describe(include="all"))
```

```python
# Checking missing values in each column

print("Missing values:")
print(df.isnull().sum())

# Calculating total missing values

total_missing = df.isnull().sum().sum()

print("\nTotal missing values:", total_missing)
```

```python
# Checking duplicate records

duplicate_count = df.duplicated().sum()

print("Total duplicate records:", duplicate_count)
```

```python
# Data Cleaning
# Removing duplicate records

df = df.drop_duplicates()

print("Duplicate records removed successfully!")

print("Updated dataset shape:", df.shape)
```

```python
# Filling missing values in categorical columns

text_columns = [
    "director",
    "cast",
    "country",
    "rating",
    "duration",
    "date_added"
]

for column in text_columns:
    df[column] = df[column].fillna("Not Available")

print("Missing values handled successfully!")
```

```python
# Removing extra spaces from text columns

for column in df.select_dtypes(include="object").columns:
    df[column] = df[column].str.strip()

print("Extra spaces removed successfully!")
```

```python
# Converting release year into numeric format

df["release_year"] = pd.to_numeric(
    df["release_year"],
    errors="coerce"
)

# Converting date_added into date format

df["date_added"] = pd.to_datetime(
    df["date_added"],
    errors="coerce"
)

print("Data types updated successfully!")

print(df.dtypes)
```

```python
# Displaying cleaned dataset

print("Cleaned dataset:")
display(df.head())

# Checking missing values again

print("\nMissing values after cleaning:")
print(df.isnull().sum())

# Checking duplicate records again

print("\nDuplicate records after cleaning:")
print(df.duplicated().sum())

# Checking final dataset shape

print("\nFinal dataset shape:", df.shape)
```

```python
# Saving the cleaned dataset

df.to_csv("cleaned_netflix_data.csv", index=False)

print("Cleaned dataset saved successfully!")
```
